```python
import numpy as np
import pandas as pd

# Data load

df = pd.read_csv('Pakwheel car data.csv')
print("Original Data Sample:")
print(df.head())

```

```python


# Data Cleaning Functions


def clean_price(price_val):
    """
    Converts 75.5 -> 7,550,000
    """
    try:
        val = float(price_val)
        return val * 100000
    except:
        return None

def clean_cc(cc_text):
    """
    Handles 'cc' and removes 'kwh' (Electric cars) to prevent errors.
    """
    if isinstance(cc_text, str):
        cc_text = cc_text.lower().strip()
        
        
        if 'kwh' in cc_text:
            return None
            
        cc_text = cc_text.replace('cc', '').strip()
        
        try:
            return float(cc_text)
        except ValueError:
            return None
            
    return float(cc_text)

def clean_km(km_text):
    if isinstance(km_text, str):
        km_text = km_text.lower().replace('km', '').replace(',', '').strip()
        try:
            return float(km_text)
        except ValueError:
            return None
    return float(km_text)

# Clean data

print("\nCleaning data...")
df['prices'] = df['prices'].apply(clean_price)
df['CC'] = df['CC'].apply(clean_cc)
df['Km_Driven'] = df['Km_Driven'].apply(clean_km)

# IMPORTANT: Drop rows with None values (This removes the 'kwh' cars automatically)
initial_count = len(df)
df.dropna(subset=['prices', 'CC', 'Km_Driven'], inplace=True)
dropped_count = initial_count - len(df)
print(f"Dropped {dropped_count} rows (mostly electric cars or bad data).")

print("Cleaned Data Sample:")
print(df[['Model', 'CC', 'prices']].head())
```

```python

df.to_csv("Cleaned_Data_Final.csv")
```
